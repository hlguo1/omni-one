## P&L walk: GMV − fees − ship − COGS = profit

```python
gmv, fees, ship, cogs = 274.00, 16.71, 24.00, 64.20
net = gmv - fees - ship
profit = net - cogs
print(f'GMV {gmv} - fees {fees} - ship {ship} = net {net:.2f}')
print(f'net {net:.2f} - COGS {cogs} = profit {profit:.2f} ({profit/gmv*100:.1f}%)')
```

## Reconcile: 6 steps

```python
steps = ['1. GMV vs channel payouts (stripe/paypal/square)', '2. fees vs channel schedule', '3. shipping vs labels', '4. COGS vs invoices (catalog)', '5. refunds vs DM/review log', '6. file close.json with month']
print('\n'.join(steps))
```
