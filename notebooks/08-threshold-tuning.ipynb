# 08 — Threshold tuning (knobs, not vibes)

All Seller OS numbers come from one table: `SCENARIO_THRESHOLDS` + `COVER_DAYS`. Change one knob at a time.

```python
THRESHOLDS = {'stockout_days': 5.0, 'cover_days': 14, 'fee_creep_pct': 12.0, 'shipping_loss_pct': 20.0, 'margin_low_pct': 10.0, 'dm_backlog_count': 3, 'price_mismatch_pct': 5.0}
print(THRESHOLDS)
```

## Stockout: days_left = on_hand / (sold_7d / 7)

```python
on_hand, sold_7d = 2, 6
vel = sold_7d/7
days = on_hand/vel
print(f'velocity={vel:.2f}/d days_left={days:.1f} -> ORDER TODAY' if days < 3 else 'watch')
```

## Cover: reorder_qty = ceil(velocity × 14 − on_hand)

```python
import math
print('reorder:', max(0, math.ceil(vel*14 - on_hand)))
```

## Fee creep: fees/GMV >= 12%

```python
gmv, fees = 274.00, 16.71
print(f'{fees/gmv*100:.1f}%', 'healthy' if fees/gmv < 0.12 else 'CREEP')
```

```python
# creep example: amazon-heavy shop
gmv, fees = 500.00, 75.00
print(f'{fees/gmv*100:.1f}%', 'healthy' if fees/gmv < 0.12 else 'CREEP — raise price or shift channel')
```

## Shipping loss per order >= 20%

```python
orders = [('Tote', 56.0, 4.0), ('Mug', 22.0, 3.5), ('Print', 15.0, 5.0)]
for p,g,s in orders:
    print(p, f'{s/g*100:.0f}%', 'LOSS' if s/g >= 0.20 else 'ok')
```

## Margin dip < 10%

```python
gmv, profit = 274.00, 169.09
print(f'margin {profit/gmv*100:.1f}%', 'healthy')
```

## DM backlog >= 3, mismatch >= 5%

```python
dms = 4
print('DM backlog' if dms >= 3 else 'inbox zero')
shopify, etsy = 28.00, 34.16
print(f'mismatch {(etsy-shopify)/shopify*100:.0f}%', 'ALIGN' if abs(etsy-shopify)/shopify >= 0.05 else 'ok')
```
